```python
import cv2
import os 
import uuid
import time
```

```python
IMAGES_PATH = 'collectimages'
labels = ['phone', 'bottle', 'pen']
number_imgs = 10
```

```python
for label in labels:
    !mkdir {'collectimages\\'+label}
    cap = cv2.VideoCapture(0)
    print('Image for {}'.format(label))
    time.sleep(3)
    for imgnum in range(number_imgs):
        ret, frame = cap.read()
        imagename = os.path.join(IMAGES_PATH, label, label+'.'+'{}.jpg'.format(str(uuid.uuid1())))
        cv2.imwrite(imagename, frame)
        # cv2.imshow('frame', frame)
        time.sleep(3)
        
    cap.release()
    
```
