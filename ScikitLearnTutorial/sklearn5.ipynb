```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

x = [4,5,6,7,8,2,3,4, 10,12]
y = [21,41,45,20,21,34,23,23, 30,22]

plt.scatter(x,y)
plt.show()
```

```python
data = list(zip(x,y))
mydata = []

for i in range(1,11):
    kmean = KMeans(n_clusters=i)
    kmean.fit(data)
    mydata.append(kmean.inertia_)

plt.plot(range(1,11), mydata, marker ='*')

plt.show()
```

```python
kmean2 = KMeans(n_clusters=2)
kmean2.fit(data)

plt.scatter(x,y, c = kmean2.labels_)
plt.show()
```
